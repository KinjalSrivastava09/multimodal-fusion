```python
# Show observed metrics, confusion matrix, and generated figures inline.
import json
from IPython.display import display, Image, Markdown
from pathlib import Path
m = json.loads(Path("results/metrics.json").read_text())["models"]
display(Markdown("## Held-out test accuracy (450 synthetic examples)"))
for k, label in [("image", "Image-only"), ("text", "Text-only"), ("fusion", "Fusion"), ("fusion_no_text", "Fusion with text embedding zeroed")]:
    print(f"{label:36s} {100*m[k]['accuracy']:.2f}%  (macro-F1 {m[k]['f1_macro']:.3f})")
for name in ["paired_examples.png", "architecture.png", "training_loss.png", "accuracy.png", "confusion_matrix.png", "latent_pca.png"]:
    display(Markdown(f"### {name.replace('_', ' ').removesuffix('.png').title()}"))
    display(Image(filename=str(Path("figures")/name)))
```
